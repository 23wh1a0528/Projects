```python
df = df.fillna(0)
df.isnull().sum()
```

```python
india_df = df[df["country"] == "India"]
india_df.head()
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(10,5))
plt.plot(india_df["year"], india_df["co2"])
plt.xlabel("Year")
plt.ylabel("CO2 Emission")
plt.title("CO2 trend for india over the years")
plt.show()
```

```python
countries = ["India", "China", "United States"]

compare_df = df[df["country"].isin(countries)]

import matplotlib.pyplot as plt

plt.figure(figsize=(12,6))  

for country in countries:
    country_data = compare_df[compare_df["country"] == country]
    
    plt.plot(country_data["year"], country_data["co2"], label=country)

plt.xlabel("Year")
plt.ylabel("CO2 Emissions")
plt.title("CO₂ Emissions Comparison: India vs China vs USA")
plt.legend()
plt.show()
```

# Result
- China’s rapid increase
- USA’s peak and decline
- India’s steady rise

```python
import matplotlib.pyplot as plt
latest_year = df["year"].max()
latest = df[df["year"] == latest_year]
top10 = latest.sort_values("co2", ascending=False).head(10)
plt.figure(figsize=(12,6))
plt.bar(top10["country"], top10["co2"])

plt.xlabel("Country")
plt.ylabel("CO2 Emissions")
plt.title(f"Top 10 CO₂ Emitting Countries in {latest_year}")
plt.xticks(rotation=45)
plt.show()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12,6))

df.boxplot(column="co2", by="country", showfliers=False)
plt.xticks([], []) 
plt.ylabel("CO₂ Emissions")
plt.title("CO₂ Distribution Across All Countries")
plt.suptitle("")
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
top20 = (
    df.groupby("country")["co2"]
    .sum()
    .sort_values(ascending=False)
    .head(20)
    .index
)

heat_df = df[df["country"].isin(top20)]
heat_pivot = heat_df.pivot(index="year", columns="country", values="co2")

plt.figure(figsize=(14,8))
sns.heatmap(heat_pivot, cmap="viridis", cbar=True)
plt.title("CO₂ Heatmap for Top 20 Countries")
plt.xlabel("Country")
plt.ylabel("Year")
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

numeric_df = df.select_dtypes(include=['float64', 'int64'])

corr_matrix = numeric_df.corr()

plt.figure(figsize=(14,10))

sns.heatmap(
    corr_matrix, 
    cmap="coolwarm",  
    linewidths=0.5, 
    annot=False  
)

plt.title("Correlation Heatmap of CO₂ Dataset")
plt.show()
```

# Interpretation

- The correlation heatmap shows that CO₂ emissions strongly correlate with GDP, energy consumption per capita, and fossil fuel-based emissions (coal, oil, gas). 
- Per-capita variables also form a strong cluster, indicating that higher standards of living lead to higher per-person emissions. 
- Fossil fuel sources are highly inter-correlated, meaning countries relying on one fossil fuel typically rely on others as well. 
- Cumulative emissions strongly correlate with cumulative fossil fuel usage, reflecting historical industrialization.

```python
import matplotlib.pyplot as plt

# Plot CO₂ per person for India 
plt.figure(figsize=(10,5))

plt.plot(india_df["year"], india_df["co2_per_capita"])

plt.xlabel("Year")                      
plt.ylabel("CO₂ per capita (tons)")    
plt.title("CO₂ Emissions Per Capita for India Over the Years")

plt.show()
```

```python
import matplotlib.pyplot as plt

# Scatter plot to show the relationship between energy usage and CO₂ emissions for India
plt.figure(figsize=(10,6))

plt.scatter(india_df["energy_per_capita"], india_df["co2_per_capita"])

plt.xlabel("Energy Consumption Per Capita")          
plt.ylabel("CO₂ Emissions Per Capita")              
plt.title("Energy Consumption vs CO₂ Emissions (India)")

plt.show()
```

# Observation
- A steady upward slope
    - As India developed, energy use per capita increased.
- A direct relationship between energy and CO₂
    - Because India's energy is still dominated by coal.
- No major decoupling yet
    - India has not reached a stage where energy use rises but CO₂ does not.

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12,6))

# Plot coal CO₂ emissions
plt.plot(india_df["year"], india_df["coal_co2"], label="Coal CO₂")

# Plot oil CO₂ emissions
plt.plot(india_df["year"], india_df["oil_co2"], label="Oil CO₂")

# Plot gas CO₂ emissions
plt.plot(india_df["year"], india_df["gas_co2"], label="Gas CO₂")

plt.xlabel("Year")
plt.ylabel("CO₂ Emissions (Million Tons)")
plt.title("Fossil Fuel CO₂ Emissions Over Time (India)")
plt.legend()

plt.show()
```

# Interpretation
- India’s fossil fuel CO₂ emissions are dominated by coal, which is significantly higher than oil or gas emissions. 
- Coal use increased steeply after 2000 due to expansion of coal-based power plants. 
- Oil emissions grew steadily because of rising transportation demand, while gas emissions remain the lowest among the three. 
- The strong growth in coal emissions indicates India’s dependence on coal for electricity generation.

```python
latest_year = df["year"].max()
latest = df[df["year"] == latest_year]
ranking = latest.sort_values("co2", ascending=False)
ranking.head(20)

india_row = ranking[ranking["country"] == "India"]
india_row
```

# Interpretation:
- India is currently the 3rd largest emitter of CO₂ globally, after China and the United States. 
- However, when emissions are measured on a per-person basis, India ranks much lower, around 120–150, indicating relatively low per-capita emissions. - Coal-based emissions dominate India’s carbon footprint, placing the country among the top coal CO₂ emitters. 
- Historically, India’s cumulative emissions are lower than long-industrialized nations such as the USA and European countries.

```python
df_sorted = df.sort_values(["country", "year"])

# Calculate year-over-year CO₂ change for each country
df_sorted["co2_yoy_change"] = df_sorted.groupby("country")["co2"].diff()

df_sorted.head(200)
```

```python
import pandas as pd
import matplotlib.pyplot as plt

heatmap_df = df_sorted.pivot_table(
    index="country",
    columns="year",
    values="co2_yoy_change"
)

plt.figure(figsize=(15, 10))

plt.imshow(heatmap_df, aspect="auto")

plt.colorbar(label="Year-over-Year CO₂ Change") 
plt.xlabel("Year")
plt.ylabel("Country")
plt.title("Heatmap of CO₂ Year-over-Year Change for All Countries")

plt.show()
```

## Final Conclusions
- Global CO₂ emissions are rising overall, driven mainly by Asia.
- Many Western countries have declining emissions, signaling successful renewable policies.
- Developing economies show rapid growth in CO₂, with big YOY jumps.
